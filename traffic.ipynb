```python
import cv2
import numpy as np
import time
import os
from ultralytics import YOLO
import torch
from google.colab.patches import cv2_imshow

class RedLightViolationDetector:
    def __init__(self, model_path="yolov8n.pt", traffic_light_model_path="yolov8n.pt"):

        self.vehicle_model = YOLO(model_path)
        self.traffic_light_model = YOLO(traffic_light_model_path)
        self.vehicle_classes = [2, 3, 5, 7]
        self.traffic_light_class = 9
        self.tracks = {}
        self.next_track_id = 1
        self.violation_line = None
        self.light_state = "unknown"  # "red", "yellow", "green", "unknown"
        self.light_state_changed_time = time.time()
        self.traffic_light_roi = None
        self.violations = []
        self.font = cv2.FONT_HERSHEY_SIMPLEX
        self.font_scale = 0.7
        self.font_thickness = 2
        self.red = (0, 0, 255)
        self.green = (0, 255, 0)
        self.yellow = (0, 255, 255)
        self.blue = (255, 0, 0)
        self.white = (255, 255, 255)

    def set_violation_line(self, y_position, frame_width):
        """Set the violation line position."""
        self.violation_line = [(0, y_position), (frame_width, y_position)]

    def set_traffic_light_roi(self, x1, y1, x2, y2):
        """Set region of interest for traffic light detection."""
        self.traffic_light_roi = (x1, y1, x2, y2)

    def detect_traffic_light_state(self, frame):
        """Detect the current state of the traffic light."""
        if self.traffic_light_roi is None:
            return "unknown"


        x1, y1, x2, y2 = self.traffic_light_roi
        roi = frame[y1:y2, x1:x2]

        if roi.size == 0:
            return self.light_state


        results = self.traffic_light_model(roi, classes=[self.traffic_light_class])

        if len(results[0].boxes) == 0:
            return self.light_state


        boxes = results[0].boxes
        areas = (boxes.xyxy[:, 2] - boxes.xyxy[:, 0]) * (boxes.xyxy[:, 3] - boxes.xyxy[:, 1])
        largest_idx = int(torch.argmax(areas))
        box = boxes.xyxy[largest_idx].cpu().numpy()


        x1_tl, y1_tl, x2_tl, y2_tl = map(int, box)
        traffic_light = roi[y1_tl:y2_tl, x1_tl:x2_tl]

        if traffic_light.size == 0:
            return self.light_state


        height = traffic_light.shape[0]


        top_section = traffic_light[0:height//3, :]
        middle_section = traffic_light[height//3:2*height//3, :]
        bottom_section = traffic_light[2*height//3:, :]


        top_hsv = cv2.cvtColor(top_section, cv2.COLOR_BGR2HSV)
        middle_hsv = cv2.cvtColor(middle_section, cv2.COLOR_BGR2HSV)
        bottom_hsv = cv2.cvtColor(bottom_section, cv2.COLOR_BGR2HSV)


        lower_red1 = np.array([0, 100, 100])
        upper_red1 = np.array([10, 255, 255])
        lower_red2 = np.array([160, 100, 100])
        upper_red2 = np.array([180, 255, 255])


        lower_yellow = np.array([20, 100, 100])
        upper_yellow = np.array([30, 255, 255])


        lower_green = np.array([40, 100, 100])
        upper_green = np.array([80, 255, 255])


        red_mask1 = cv2.inRange(top_hsv, lower_red1, upper_red1)
        red_mask2 = cv2.inRange(top_hsv, lower_red2, upper_red2)
        red_pixels = cv2.countNonZero(red_mask1) + cv2.countNonZero(red_mask2)


        yellow_mask = cv2.inRange(middle_hsv, lower_yellow, upper_yellow)
        yellow_pixels = cv2.countNonZero(yellow_mask)


        green_mask = cv2.inRange(bottom_hsv, lower_green, upper_green)
        green_pixels = cv2.countNonZero(green_mask)


        max_pixels = max(red_pixels, yellow_pixels, green_pixels)

        if max_pixels < 10:
            return self.light_state

        if red_pixels == max_pixels:
            new_state = "red"
        elif yellow_pixels == max_pixels:
            new_state = "yellow"
        elif green_pixels == max_pixels:
            new_state = "green"
        else:
            new_state = self.light_state


        if new_state != self.light_state:
            self.light_state_changed_time = time.time()

        self.light_state = new_state
        return self.light_state

    def simple_tracking(self, detections, iou_threshold=0.3):
        """
        Simple object tracking based on IoU between frames
        """
        new_tracks = {}
        used_detections = set()


        for track_id, track_info in self.tracks.items():
            last_bbox = track_info['bbox']
            best_iou = 0
            best_detection_idx = -1


            for i, bbox in enumerate(detections):
                if i in used_detections:
                    continue

                iou = self.calculate_iou(last_bbox, bbox)
                if iou > best_iou and iou > iou_threshold:
                    best_iou = iou
                    best_detection_idx = i


            if best_detection_idx >= 0:
                new_tracks[track_id] = {
                    'bbox': detections[best_detection_idx],
                    'age': track_info['age'] + 1,
                    'missed': 0
                }
                used_detections.add(best_detection_idx)
            else:

                if track_info['missed'] < 5:
                    new_tracks[track_id] = {
                        'bbox': last_bbox,
                        'age': track_info['age'] + 1,
                        'missed': track_info['missed'] + 1
                    }


        for i, bbox in enumerate(detections):
            if i not in used_detections:
                new_tracks[self.next_track_id] = {
                    'bbox': bbox,
                    'age': 1,
                    'missed': 0
                }
                self.next_track_id += 1

        self.tracks = new_tracks
        return self.tracks

    def calculate_iou(self, box1, box2):
        """Calculate IoU between two bounding boxes."""
        x1_1, y1_1, x2_1, y2_1 = box1
        x1_2, y1_2, x2_2, y2_2 = box2

        x_left = max(x1_1, x1_2)
        y_top = max(y1_1, y1_2)
        x_right = min(x2_1, x2_2)
        y_bottom = min(y2_1, y2_2)

        if x_right < x_left or y_bottom < y_top:
            return 0.0

        intersection_area = (x_right - x_left) * (y_bottom - y_top)


        box1_area = (x2_1 - x1_1) * (y2_1 - y1_1)
        box2_area = (x2_2 - x1_2) * (y2_2 - y1_2)
        union_area = box1_area + box2_area - intersection_area


        if union_area <= 0:
            return 0.0

        return intersection_area / union_area

    def detect_violations(self, frame):
        """Process a video frame to detect red light violations."""

        display_frame = frame.copy()
        frame_height, frame_width = frame.shape[:2]


        if self.violation_line is None:
            self.set_violation_line(int(frame_height * 0.7), frame_width)


        if self.traffic_light_roi is None:

            self.set_traffic_light_roi(
                frame_width // 2, 0,
                frame_width, frame_height // 3
            )


        traffic_light_state = self.detect_traffic_light_state(frame)


        cv2.rectangle(
            display_frame,
            (self.traffic_light_roi[0], self.traffic_light_roi[1]),
            (self.traffic_light_roi[2], self.traffic_light_roi[3]),
            self.blue, 2
        )


        results = self.vehicle_model(frame, classes=self.vehicle_classes)


        detections = []
        if len(results[0].boxes) > 0:
            for box in results[0].boxes.xyxy.cpu().numpy():
                x1, y1, x2, y2 = map(int, box)
                detections.append((x1, y1, x2, y2))

        tracks = self.simple_tracking(detections)


        line_color = self.red if traffic_light_state == "red" else (
            self.yellow if traffic_light_state == "yellow" else self.green
        )
        cv2.line(display_frame, self.violation_line[0], self.violation_line[1], line_color, 2)


        cv2.putText(
            display_frame,
            f"Light: {traffic_light_state.upper()}",
            (frame_width - 200, 30),
            self.font, self.font_scale, line_color, self.font_thickness
        )


        for track_id, track_info in tracks.items():
            if track_info['missed'] > 0:
                continue


            x1, y1, x2, y2 = track_info['bbox']


            cv2.rectangle(display_frame, (x1, y1), (x2, y2), self.green, 2)

            vehicle_bottom_center = (int((x1 + x2) / 2), y2)


            if (vehicle_bottom_center[1] >= self.violation_line[0][1] and
                traffic_light_state == "red"):


                cv2.rectangle(display_frame, (x1, y1), (x2, y2), self.red, 3)


                if not any(v['track_id'] == track_id for v in self.violations):
                    violation = {
                        'track_id': track_id,
                        'time': time.time(),
                        'position': vehicle_bottom_center,
                        'bbox': (x1, y1, x2, y2)
                    }
                    self.violations.append(violation)


                    violation_img = frame.copy()
                    cv2.rectangle(violation_img, (x1, y1), (x2, y2), self.red, 3)
                    cv2.putText(
                        violation_img,
                        f"RED LIGHT VIOLATION ID: {track_id}",
                        (x1, y1 - 10),
                        self.font, self.font_scale, self.red, self.font_thickness
                    )


                    os.makedirs("violations", exist_ok=True)
                    cv2.imwrite(f"violations/violation_{int(time.time())}_{track_id}.jpg", violation_img)


            cv2.putText(
                display_frame,
                f"ID: {track_id}",
                (x1, y1 - 10),
                self.font, self.font_scale, self.white, self.font_thickness
            )


        cv2.putText(
            display_frame,
            f"Violations: {len(self.violations)}",
            (20, 30),
            self.font, self.font_scale, self.red, self.font_thickness
        )

        return display_frame, self.violations

    def process_video(self, video_path, output_path=None, display=False):
        """Process a video file and detect violations."""
        cap = cv2.VideoCapture(video_path)


        if not cap.isOpened():
            print(f"Error: Could not open video {video_path}")
            return


        frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        fps = cap.get(cv2.CAP_PROP_FPS)


        out = None
        if output_path:
            fourcc = cv2.VideoWriter_fourcc(*'mp4v')
            out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))


        frame_count = 0
        max_display_frames = 5

        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break


            processed_frame, _ = self.detect_violations(frame)


            if out:
                out.write(processed_frame)


            if display and frame_count % int(fps) == 0 and frame_count // int(fps) < max_display_frames:
                print(f"Sample frame at {frame_count // int(fps)} seconds:")
                cv2_imshow(processed_frame)

            frame_count += 1


            if frame_count % 100 == 0:
                print(f"Processed {frame_count} frames")


        cap.release()
        if out:
            out.release()

        print(f"Video processing complete. Total frames: {frame_count}")

        return self.violations
if __name__ == "__main__":

    detector = RedLightViolationDetector()


    violations = detector.process_video("/content/traffic2.mp4", "output_video.mp4", display=True)


    print(f"Total violations detected: {len(violations)}")
    for v in violations:
        print(f"Vehicle ID {v['track_id']} violation at time {v['time']}")
```
